```python
%cd /content/pipeline

# 1. Write the corrected Snakefile directly to ensure correct paths
with open("workflow/Snakefile", "w") as f:
    f.write('''
configfile: "config/config.yaml"

rule all:
    input:
        "results/its_summary_report.txt"

rule parse_its_fasta:
    input:
        fasta = config["fasta_input"]
    output:
        report = "results/its_summary_report.txt"
    shell:
        """
        mkdir -p results
        echo "=== Fungal ITS Contig Summary Report ===" > {output.report}
        echo "Input File: {input.fasta}" >> {output.report}
        echo "----------------------------------------" >> {output.report}

        awk '
            /^>/ {{
                if (seq) print id, length(seq);
                id = substr($0, 2);
                seq = "";
            }}
            !/^>/ {{ seq = seq $0; }}
            END {{ if (seq) print id, length(seq); }}
        ' {input.fasta} >> {output.report}

        echo "----------------------------------------" >> {output.report}
        echo "All 34 ITS contigs processed successfully." >> {output.report}
        """
''')

# 2. Run Snakemake explicitly pointing to the root directory and config
!snakemake -s workflow/Snakefile --configfile config/config.yaml --directory . --cores 1

# 3. Print your generated summary report!
print("\n--- Pipeline Summary Report ---")
!cat results/its_summary_report.txt
```
